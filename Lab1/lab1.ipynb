3.6 Separate the data into 3 sub-frames called “train”, “test” and “validate”. The first set should contain
around 60% of the entire dataset (first 60% of the rows) while the second and the third must each contain
around 20% of the data. Check if the difference in mean between the train and the validate datasets is
smaller than the standard deviation of the test dataset. You can use any numerical feature for computing
the means (eg. Height). Plot the histograms for the 3 dataset distributions.

```python
# -------------------------------------------------------------
# 1. Split into train (60%), test (20%), and validate (20%)
# -------------------------------------------------------------
n = len(data)
train_end = int(0.6 * n)   # 60%
test_end = int(0.8 * n)    # 60% + 20% = 80%

train = data.iloc[:train_end]
test = data.iloc[train_end:test_end]
validate = data.iloc[test_end:]

print(f"Total rows: {n}")
print(f"Train set:    {len(train)} rows ({len(train)/n*100:.0f}%)")
print(f"Test set:     {len(test)} rows ({len(test)/n*100:.0f}%)")
print(f"Validate set: {len(validate)} rows ({len(validate)/n*100:.0f}%)\n")

# -------------------------------------------------------------
# 2. Check: |mean(train) - mean(validate)| < std(test)
# -------------------------------------------------------------
feature = 'Height'

mean_train = train[feature].mean()
mean_val = validate[feature].mean()
diff_mean = abs(mean_train - mean_val)

std_test = test[feature].std()

print(f"Mean {feature} (Train):     {mean_train:.4f}")
print(f"Mean {feature} (Validate):  {mean_val:.4f}")
print(f"Difference in means:        {diff_mean:.4f}")
print(f"Std dev (Test):             {std_test:.4f}")

condition_met = diff_mean < std_test
print(f"\nIs difference in means ({diff_mean:.4f}) < std dev of test ({std_test:.4f})? -> {condition_met}")

# -------------------------------------------------------------
# 3. Plot histograms for the 3 distributions
# -------------------------------------------------------------
plt.figure(figsize=(15, 4))

# Train histogram
plt.subplot(1, 3, 1)
plt.hist(train[feature].dropna(), bins=30, color='skyblue', edgecolor='black')
plt.title(f'Train Set ({feature})')
plt.xlabel(feature)
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.5)

# Test histogram
plt.subplot(1, 3, 2)
plt.hist(test[feature].dropna(), bins=30, color='salmon', edgecolor='black')
plt.title(f'Test Set ({feature})')
plt.xlabel(feature)
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.5)

# Validate histogram
plt.subplot(1, 3, 3)
plt.hist(validate[feature].dropna(), bins=30, color='lightgreen', edgecolor='black')
plt.title(f'Validate Set ({feature})')
plt.xlabel(feature)
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
